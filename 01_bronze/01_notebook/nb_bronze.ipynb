### Customer_Survey Table Bronze injestion

```python
%sql
-- RAW
CREATE SCHEMA IF NOT EXISTS automobilerepair.bronze;

CREATE OR REPLACE TABLE automobilerepair.bronze.raw_customer_survey AS
SELECT * FROM automobilerepair.azure_blob_storage.customer_survey;
```

```python
%sql
-- STAGING with SCD Type 1 (in bronze schema)
CREATE OR REPLACE TABLE automobilerepair.bronze.stg_customer_survey AS
SELECT 
    survey_id,
    order_id,
    survey_sent_date,
    survey_response_date,
    responded_flag,
    delivered_on_time_rating,
    work_quality_rating,
    cleanliness_rating,
    communication_rating,
    overall_satisfaction_rating,
    _modified,
    current_timestamp() AS inserted_at,
    current_timestamp() AS updated_at
FROM automobilerepair.bronze.raw_customer_survey;
```
