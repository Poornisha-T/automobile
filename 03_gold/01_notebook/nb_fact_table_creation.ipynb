```python
%sql
CREATE OR REPLACE TABLE automobilerepair.gold.fact_orderlifecycle AS
SELECT
    -- Surrogate key to handle one-to-many relationship
    ROW_NUMBER() OVER (ORDER BY o.order_id, e.estimate_id) AS orderlifecycle_id,
    
    -- All columns from orders
    o.order_id,
    o.customer_id,
    o.store_id,
    o.technician_id,
    o.service_type,
    o.vehicle_in_datetime,
    o.vehicle_out_datetime,
    o.planned_work_start_datetime,
    o.actual_work_start_datetime,
    o.planned_completion_datetime,
    o.vehicle_id,
    o.actual_completion_datetime,
    o.promised_delivery_datetime,
    o.actual_delivery_datetime,
    o.order_status,
    
    i.invoice_id,
    i.invoice_date,
    i.invoice_amount,
    i.payment_mode,
    i.currency,
    
    e.estimate_id,
    e.version_no,
    e.estimate_amount,

    o.inserted_at,
    o.updated_at
    
FROM automobilerepair.silver.slv_order o
LEFT JOIN automobilerepair.silver.slv_invoice i ON o.order_id = i.order_id
LEFT JOIN automobilerepair.silver.slv_estimate e ON o.order_id = e.order_id;
```

```python
%sql
CREATE OR REPLACE TABLE  automobilerepair.gold.fact_budget AS
SELECT ns_store_id as store_id, month, budget_amount, inserted_at, updated_at
FROM automobilerepair.silver.slv_budget;
```
