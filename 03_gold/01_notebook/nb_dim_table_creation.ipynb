```python
%sql
CREATE OR REPLACE TABLE gold.dim_date AS
SELECT DISTINCT
    invoice_date AS date,
    YEAR(invoice_date) AS year,
    MONTH(invoice_date) AS month,
    DAY(invoice_date) AS day,
    WEEKOFYEAR(invoice_date) AS week
FROM silver.slv_invoice;
```
