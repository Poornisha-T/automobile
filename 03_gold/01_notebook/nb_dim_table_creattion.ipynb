```python
%sql
CREATE OR REPLACE TABLE gold.dim_customer AS
SELECT
    customer_id,
    customer_name,
    customer_phone
FROM silver.slv_customer;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_vehicle AS
SELECT
    vehicle_id,
    vehicle_no,
    vehicle_make,
    vehicle_model
FROM silver.slv_vehicle;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_staff AS
SELECT
    technician_id AS staff_id,
    technician_name
FROM silver.slv_technician;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_store AS
SELECT *
FROM silver.slv_store;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_date AS
SELECT DISTINCT
    invoice_date AS date,
    YEAR(invoice_date) AS year,
    MONTH(invoice_date) AS month,
    DAY(invoice_date) AS day,
    WEEKOFYEAR(invoice_date) AS week
FROM silver.slv_invoice;
```

```python
%sql
CREATE OR REPLACE TABLE gold.fact_orders AS
SELECT
    order_id,
    customer_id,
    vehicle_id,
    technician_id AS staff_id,
    store_id,

    vehicle_in_datetime,
    vehicle_out_datetime,

    repair_duration,
    delay_flag,

    promised_delivery_datetime,
    actual_delivery_datetime

FROM silver.slv_order;
```

```python
%sql
CREATE OR REPLACE TABLE gold.fact_revenue AS
SELECT
    invoice_id,
    order_id,
    invoice_amount,
    invoice_date,
    payment_mode,
    currency
FROM silver.slv_invoice;
```

```python
%sql
CREATE OR REPLACE TABLE gold.fact_budget AS
SELECT
    ns_store_id AS store_id,
    month,
    budget_amount
FROM silver.slv_ns_budget;
```
